```python
import pandas as pd
import numpy as np

print("=" * 60)
print("Excel Data Transformation Script - CANDIDATE ONLY")
print("=" * 60)

# Read Excel file
print("\nReading Excel file...")
candidate = pd.read_excel(base_dir / "candidate_cleaned_no_outliers_k3.xlsx")
```

```python
# STEP 1: Remove birthdate and address columns from CANDIDATE
print("\n" + "="*60)
print("STEP 1: Removing birthdate and address columns from CANDIDATE")
print("="*60)

columns_to_remove = [
    'C_ADDRESS', 
    'DATE_OF_BIRTH',
]

columns_found = [col for col in columns_to_remove if col in candidate.columns]

print(f"\nColumns to remove: {columns_found}")

if columns_found:
    candidate = candidate.drop(columns=columns_found)
    print(f"✓ Removed {len(columns_found)} columns")
else:
    print("No matching columns found to remove")

print(f"\nRemaining columns in CANDIDATE: {len(candidate.columns)}")
```

```python
# STEP 2: Transform qualitative data to quantitative
print("\n" + "="*60)
print("STEP 2: Transforming qualitative to quantitative data")
print("="*60)

# GENDER (CANDIDATE)
print("\n--- GENDER (CANDIDATE) ---")
if 'GENDER' in candidate.columns:
    print(f"Unique values: {candidate['GENDER'].unique()}")
    print(f"Value counts:")
    print(candidate['GENDER'].value_counts(dropna=False))
    
    gender_mapping = {
        'MALE': 1,
        'FEMALE': 2,
        'NONBINARY': 3,
        'DECLINE': 0
    }
    
    candidate['GENDER_NUMERIC'] = candidate['GENDER'].map(gender_mapping)
    # Blanks remain as NaN
    
    print("✓ GENDER transformed")
    print(f"Mapping preview:")
    print(candidate[['GENDER', 'GENDER_NUMERIC']].drop_duplicates().sort_values('GENDER_NUMERIC'))
else:
    print("GENDER column not found")
```

```python
# LAST_YEAR_INCOME (CANDIDATE)
print("\n--- LAST_YEAR_INCOME (CANDIDATE) ---")
if 'LAST_YEAR_INCOME' in candidate.columns:
    print(f"Unique values: {candidate['LAST_YEAR_INCOME'].unique()}")
    print(f"Value counts:")
    print(candidate['LAST_YEAR_INCOME'].value_counts(dropna=False))
    
    income_mapping = {
        'UNDER_5000': 1,
        'R_5000_9999': 2,
        'R_10000_19999': 3,
        'R_20000_29999': 4,
        'R_30000_39999': 5,
        'R_40000_49999': 6,
        'R_50000_74999': 7,
        'R_75000_ABOVE': 8
    }
    
    candidate['LAST_YEAR_INCOME_NUMERIC'] = candidate['LAST_YEAR_INCOME'].map(income_mapping)
    # Blanks remain as NaN
    
    print("✓ LAST_YEAR_INCOME transformed")
    print(f"Mapping preview:")
    print(candidate[['LAST_YEAR_INCOME', 'LAST_YEAR_INCOME_NUMERIC']].drop_duplicates().sort_values('LAST_YEAR_INCOME_NUMERIC'))
else:
    print("LAST_YEAR_INCOME column not found")
```

```python
# HIGHEST_GRADE_COMPLETED (CANDIDATE)
print("\n--- HIGHEST_GRADE_COMPLETED (CANDIDATE) ---")
if 'HIGHEST_GRADE_COMPLETED' in candidate.columns:
    print(f"Unique values: {candidate['HIGHEST_GRADE_COMPLETED'].unique()}")
    print(f"Value counts:")
    print(candidate['HIGHEST_GRADE_COMPLETED'].value_counts(dropna=False))
    
    grade_mapping = {
        'NEVER': 0,
        'C_PRE_5': 5,       # Pre-kindergarten through 5th grade
        'C_6_8': 8,         # 6th through 8th grade
        'C_9': 9,           # 9th grade
        'C_10': 10,         # 10th grade
        'C_11': 11,         # 11th grade
        'C_12': 12,         # 12th grade (High School)
        'DONT_REMEMBER': -1,
        'OTHER': -2
    }
    
    candidate['HIGHEST_GRADE_COMPLETED_NUMERIC'] = candidate['HIGHEST_GRADE_COMPLETED'].map(grade_mapping)
    # Blanks remain as NaN
    
    print("✓ HIGHEST_GRADE_COMPLETED transformed")
    print(f"Mapping preview:")
    print(candidate[['HIGHEST_GRADE_COMPLETED', 'HIGHEST_GRADE_COMPLETED_NUMERIC']].drop_duplicates().sort_values('HIGHEST_GRADE_COMPLETED_NUMERIC'))
else:
    print("HIGHEST_GRADE_COMPLETED column not found")
```

```python
# TESTING_REASON (CANDIDATE)
print("\n--- TESTING_REASON (CANDIDATE) ---")
if 'TESTING_REASON' in candidate.columns:
    print(f"Unique values: {candidate['TESTING_REASON'].unique()}")
    print(f"Value counts:")
    print(candidate['TESTING_REASON'].value_counts(dropna=False))
    
    reason_mapping = {
        'WORK_RELATED': 1,
        'EDUCATIONAL_GAIN': 2,
        'PERSONAL_GAIN': 3,
        'ENTRANCE_TO_MILITARY': 4,
        'SPECIAL_REQUIREMENT': 5
    }
    
    candidate['TESTING_REASON_NUMERIC'] = candidate['TESTING_REASON'].map(reason_mapping)
    # Blanks remain as NaN
    
    print("✓ TESTING_REASON transformed")
    print(f"Mapping preview:")
    print(candidate[['TESTING_REASON', 'TESTING_REASON_NUMERIC']].drop_duplicates().sort_values('TESTING_REASON_NUMERIC'))
else:
    print("TESTING_REASON column not found")
```
