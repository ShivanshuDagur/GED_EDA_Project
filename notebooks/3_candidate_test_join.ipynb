```python
import pandas as pd
import numpy as np
```

```python
candidate_path = 'C://Users//amogh//Documents//GED_EDA_Project//data//2. Outliers+Distribution//candidate_cleaned_no_outliers_k3.xlsx'
test_path = 'C://Users//amogh//Documents//GED_EDA_Project//data//2. Outliers+Distribution//test_cleaned_no_outliers_k3.xlsx'
```

```python
candidate_df = pd.read_excel(candidate_path)
test_df = pd.read_excel(test_path)
```

```python
joint_df = test_df.merge(candidate_df, on="CANDIDATE_ID", how="inner")
```

```python
joint_df.shape
```

```python
# Sorting the joint table by EXAM_START
joint_df = joint_df.sort_values(
    by=['CANDIDATE_ID', 'EXAM_START'],
    na_position='last'
)

# Creating a unique key for every attempt made by the candidate
joint_df['CANDIDATE_ATTEMPT_ID'] = (
    joint_df['CANDIDATE_ID'].astype(str) + '_' +
    (joint_df.groupby('CANDIDATE_ID').cumcount() + 1).astype(str)
)

joint_df.insert(0, 'CANDIDATE_ATTEMPT_ID', joint_df.pop('CANDIDATE_ATTEMPT_ID'))
```

```python
assert joint_df['CANDIDATE_ATTEMPT_ID'].is_unique, "Unique key is NOT unique!"
```
