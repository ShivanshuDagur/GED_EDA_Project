```python
# Pacakage loading
from google.colab import files
import pandas as pd
import numpy as np
import re
import io
```

```python
# Load files
df = pd.read_csv('4_test_candidate_cleaned_qual_2_quant.csv', encoding='latin1').replace(['NULL', 'null', 'NA', 'No street address available', 'No city available'], np.nan)
```

```python
# Ensuring columns name does not have any white space
df.columns = df.columns.str.strip()
```

```python
# Date parsing
date_cols = ['ACCOUNT_SETUP_COMPLETE_DATE', 'CREDENTIAL_DATE']
for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')
```

```python
# Type Fixing
df['CANDIDATE_ID'] = df['CANDIDATE_ID'].astype(str)
df['RESULT_ID'] = df['RESULT_ID'].astype(str)
df['TEST_CENTER_ID'] = df['TEST_CENTER_ID'].astype(str)
```

```python
# Dealing with the freeform zip
def clean_zip(zip_code):
    # make sure zip is string and remove null data
    zip_str = str(zip_code).strip()
    if zip_str.lower() in ['nan', 'null', 'NA', '0']:
        return np.nan

    # remove all non numeric characters
    digits_only = re.sub(r'\D', '', zip_str)
    length = len(digits_only)

    # check number, remove singular and put 0
    if length <= 2:
        return np.nan
    elif length < 5:
        return digits_only.zfill(5)
    else:
        return digits_only[:5]

# apply to data
df['ZIP'] = df['ZIP'].apply(clean_zip)
df['T_POSTAL_CODE'] = df['T_POSTAL_CODE'].apply(clean_zip)
```

```python
df.to_csv('5_test_candidate_cleaned_final.csv', index=False)
```
