```python
import pandas as pd
import numpy as np

print("=" * 60)
print("Excel Data Transformation Script")
print("=" * 60)

# Read Excel file
print("\nReading Excel file...")
merged_data = pd.read_excel("C:/Users/davey/OneDrive/Desktop/3_candidate_test_attempts.xlsx")
```

```python
# STEP 1: Remove 'Civics' subject rows from merged_data
print("\n" + "="*60)
print("STEP 1: Removing 'Civics' subject rows from DATASET")
print("="*60)

if 'EXAM_SUBJECT' in merged_data.columns:
    original_rows = len(merged_data)
    print(f"\nOriginal rows: {original_rows}")
    print(f"Unique subjects before removal:")
    print(merged_data['EXAM_SUBJECT'].value_counts(dropna=False))
    
    merged_data = merged_data[merged_data['EXAM_SUBJECT'] != 'CIVICS']
    removed_rows = original_rows - len(merged_data)
    
    print(f"\nRemoved {removed_rows} rows with 'CIVICS' subject")
    print(f"Remaining rows: {len(merged_data)}")
    print(f"\nUnique subjects after removal:")
    print(merged_data['EXAM_SUBJECT'].value_counts(dropna=False))
else:
    print("EXAM_SUBJECT column not found")
```

```python
# STEP 2: Remove birthdate and address columns
print("\n" + "="*60)
print("STEP 2: Removing birthdate and address columns from DATASET")
print("="*60)

columns_to_remove = [
    'C_ADDRESS', 
    'DATE_OF_BIRTH',
]

columns_found = [col for col in columns_to_remove if col in merged_data.columns]
print(f"\nColumns to remove: {columns_found}")

if columns_found:
    merged_data = merged_data.drop(columns=columns_found)
    print(f"✓ Removed {len(columns_found)} columns")
else:
    print("No matching columns found to remove")

print(f"\nRemaining columns in DATASET: {len(merged_data.columns)}")
```

```python
# STEP 3: Transform qualitative data to quantitative
print("\n" + "="*60)
print("STEP 3: Transforming qualitative to quantitative data")
print("="*60)

# LANGUAGE_CODE
print("\n--- LANGUAGE_CODE ---")
if 'LANGUAGE_CODE' in merged_data.columns:
    print(f"Unique values: {merged_data['LANGUAGE_CODE'].unique()}")
    print(f"Value counts:")
    print(merged_data['LANGUAGE_CODE'].value_counts(dropna=False))
    
    language_mapping = {
        'ENU': 1,  # English
        'ESP': 2,  # Spanish
        'UNKNOWN': 0
    }
    
    # Replace original column
    merged_data['LANGUAGE_CODE'] = merged_data['LANGUAGE_CODE'].map(language_mapping)
    
    print("✓ LANGUAGE_CODE transformed")
    print(f"New value counts:")
    print(merged_data['LANGUAGE_CODE'].value_counts(dropna=False))
else:
    print("LANGUAGE_CODE column not found")
```

```python
# EXAM_SUBJECT
print("\n--- EXAM_SUBJECT ---")
if 'EXAM_SUBJECT' in merged_data.columns:
    print(f"Unique values: {merged_data['EXAM_SUBJECT'].unique()}")
    print(f"Value counts:")
    print(merged_data['EXAM_SUBJECT'].value_counts(dropna=False))
    
    subject_mapping = {
        'MATH': 1,
        'SCIENCE': 2,
        'REASONING': 3,
        'SOCIAL_STUDIES': 4,
        'UNKNOWN': 0
    }
    
    # Replace original column
    merged_data['EXAM_SUBJECT'] = merged_data['EXAM_SUBJECT'].map(subject_mapping)
    
    print("✓ EXAM_SUBJECT transformed")
    print(f"New value counts:")
    print(merged_data['EXAM_SUBJECT'].value_counts(dropna=False))
else:
    print("EXAM_SUBJECT column not found")
```

```python
# GENDER
print("\n--- GENDER ---")
if 'GENDER' in merged_data.columns:
    print(f"Unique values: {merged_data['GENDER'].unique()}")
    print(f"Value counts:")
    print(merged_data['GENDER'].value_counts(dropna=False))
    
    gender_mapping = {
        'MALE': 1,
        'FEMALE': 2,
        'NONBINARY': 3,
        'DECLINE': 4,
        'UNKNOWN': 0
    }
    
    # Replace original column
    merged_data['GENDER'] = merged_data['GENDER'].map(gender_mapping)
    
    print("✓ GENDER transformed")
    print(f"New value counts:")
    print(merged_data['GENDER'].value_counts(dropna=False))
else:
    print("GENDER column not found")
```

```python
# LAST_YEAR_INCOME
print("\n--- LAST_YEAR_INCOME ---")
if 'LAST_YEAR_INCOME' in merged_data.columns:
    print(f"Unique values: {merged_data['LAST_YEAR_INCOME'].unique()}")
    print(f"Value counts:")
    print(merged_data['LAST_YEAR_INCOME'].value_counts(dropna=False))
    
    income_mapping = {
        'UNDER_5000': 1,
        'R_5000_9999': 2,
        'R_10000_19999': 3,
        'R_20000_29999': 4,
        'R_30000_39999': 5,
        'R_40000_49999': 6,
        'R_50000_74999': 7,
        'R_75000_ABOVE': 8,
        'UNKNOWN': 0
    }
    
    # Replace original column
    merged_data['LAST_YEAR_INCOME'] = merged_data['LAST_YEAR_INCOME'].map(income_mapping)
    
    print("✓ LAST_YEAR_INCOME transformed")
    print(f"New value counts:")
    print(merged_data['LAST_YEAR_INCOME'].value_counts(dropna=False))
else:
    print("LAST_YEAR_INCOME column not found")
```

```python
# HIGHEST_GRADE_COMPLETED
print("\n--- HIGHEST_GRADE_COMPLETED ---")
if 'HIGHEST_GRADE_COMPLETED' in merged_data.columns:
    print(f"Unique values: {merged_data['HIGHEST_GRADE_COMPLETED'].unique()}")
    print(f"Value counts:")
    print(merged_data['HIGHEST_GRADE_COMPLETED'].value_counts(dropna=False))
    
    grade_mapping = {
        'NEVER': 9,
        'C_PRE_5': 1,       # Pre-kindergarten through 5th grade
        'C_6_8': 2,         # 6th through 8th grade
        'C_9': 3,           # 9th grade
        'C_10': 4,          # 10th grade
        'C_11': 5,          # 11th grade
        'C_12': 6,          # 12th grade (High School)
        'DONT_REMEMBER': 7,
        'OTHER': 8,
        'UNKNOWN': 0
    }
    
    # Replace original column
    merged_data['HIGHEST_GRADE_COMPLETED'] = merged_data['HIGHEST_GRADE_COMPLETED'].map(grade_mapping)
    
    print("✓ HIGHEST_GRADE_COMPLETED transformed")
    print(f"New value counts:")
    print(merged_data['HIGHEST_GRADE_COMPLETED'].value_counts(dropna=False))
else:
    print("HIGHEST_GRADE_COMPLETED column not found")
```

```python
# TESTING_REASON
print("\n--- TESTING_REASON ---")
if 'TESTING_REASON' in merged_data.columns:
    print(f"Unique values: {merged_data['TESTING_REASON'].unique()}")
    print(f"Value counts:")
    print(merged_data['TESTING_REASON'].value_counts(dropna=False))
    
    reason_mapping = {
        'WORK_RELATED': 1,
        'EDUCATIONAL_GAIN': 2,
        'PERSONAL_GAIN': 3,
        'ENTRANCE_TO_MILITARY': 4,
        'SPECIAL_REQUIREMENT': 5,
        'UNKNOWN': 0
    }
    
    # Replace original column
    merged_data['TESTING_REASON'] = merged_data['TESTING_REASON'].map(reason_mapping)
    
    print("✓ TESTING_REASON transformed")
    print(f"New value counts:")
    print(merged_data['TESTING_REASON'].value_counts(dropna=False))
else:
    print("TESTING_REASON column not found")
```

```python
# ETHNICITY
print("\n--- ETHNICITY ---")
if 'ETHNICITY' in merged_data.columns:
    print(f"Unique values: {merged_data['ETHNICITY'].unique()}")
    print(f"Value counts:")
    print(merged_data['ETHNICITY'].value_counts(dropna=False))
    
    ethnicity_mapping = {
        'DECLINE': 3,
        'NON_HISPANIC_LATINO': 1,
        'HISPANIC_LATINO': 2,
        'UNKNOWN': 0
    }
    
    # Replace original column
    merged_data['ETHNICITY'] = merged_data['ETHNICITY'].map(ethnicity_mapping)
    
    print("✓ ETHNICITY transformed")
    print(f"New value counts:")
    print(merged_data['ETHNICITY'].value_counts(dropna=False))
else:
    print("ETHNICITY column not found")
```

```python
# STEP 4: Save transformed data
print("\n" + "="*60)
print("STEP 4: Saving transformed data")
print("="*60)

output_file = "C:/Users/davey/OneDrive/Desktop/4_test_candidate_cleaned_qual_2_quant.xlsx"

with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
    merged_data.to_excel(writer, sheet_name='MERGED_DATA', index=False)

print(f"\n✓ Transformed data saved to '{output_file}'")
```
